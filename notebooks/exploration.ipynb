# Art Price Predictor — EDA
Смотрю на данные, ищу закономерности перед тем как тренировать модель.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/sample_data.csv')
print(df.shape)
df.head()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
df['price_usd'].hist(bins=40, ax=ax[0])
ax[0].set_title('price distribution')
ax[0].set_xlabel('price_usd')
np.log10(df['price_usd']).hist(bins=40, ax=ax[1], color='orange')
ax[1].set_title('log10(price)')
plt.show()
```

```python
by_artist = df.groupby('artist')['price_usd'].mean().sort_values(ascending=False).head(10)
by_artist.plot(kind='barh', figsize=(10, 5), color='steelblue')
plt.title('top 10 artists by avg price')
plt.xlabel('avg price_usd')
plt.gca().invert_yaxis()
plt.show()
```

```python
df['area'] = df['width_cm'] * df['height_cm']
df['age'] = df['sale_year'] - df['year_painted']
num_cols = ['width_cm', 'height_cm', 'area', 'year_painted', 'sale_year', 'age', 'price_usd']
corr = df[num_cols].corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('correlation matrix')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 4))
df.boxplot(column='price_usd', by='technique', ax=ax[0])
ax[0].set_yscale('log')
ax[0].set_title('price by technique (log)')
ax[0].set_xlabel('')
plt.suptitle('')

ax[1].scatter(df['area'], df['price_usd'], alpha=0.4)
ax[1].set_yscale('log')
ax[1].set_xlabel('area cm²')
ax[1].set_ylabel('price_usd (log)')
ax[1].set_title('area vs price')
plt.tight_layout()
plt.show()
```
